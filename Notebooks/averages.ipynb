```python
import pandas as pd

datas = ['magpie_1900', 'movie_pos_1900','magpie_no_boundary_1900','books_1900','trofi_1900','sens_1900','polite_1900','idem_1900']
seed = 10

from sklearn.metrics import f1_score, precision_score, recall_score, accuracy_score

table = {'train_data': [], 'category': [], 'average_f1': [], 'average_prec': [], 'average_rec': [], 'average_acc': []}

aa = pd.read_csv('en_test.csv')

for i in datas:
    for j in aa['EUPH_STATUS'].dropna().unique():
        #print(j)
        total_f1 = 0
        total_prec = 0
        total_rec = 0
        total_acc = 0
        for k in range(0,seed):
            a = pd.read_csv('tables_new/'+str(k)+'_'+i+'_boundary_table.csv')
            a = a[a['status']==j]
            total_f1 += f1_score(a['label'],a['predicted'],average='macro')
            total_prec += precision_score(a['label'],a['predicted'],average='macro')
            total_rec += recall_score(a['label'],a['predicted'],average='macro')
            total_acc += accuracy_score(a['label'],a['predicted'])
        table['train_data'].append(i)
        table['category'].append(j)
        #print(table['category'])
        table['average_f1'].append(total_f1/seed)
        table['average_prec'].append(total_prec/seed)
        table['average_rec'].append(total_rec/seed)
        table['average_acc'].append(total_acc/seed)
#print(table)
```

```python
table = pd.DataFrame(table)
table.to_csv('average_table_by_status_boundaried.csv')
```

```python
import pandas as pd

datas = ['pre','magpie_1900','trofi_boundary_1900', 'movie_pos_1900','magpie_no_boundary_1900','books_1900','trofi_1900','sens_1900','polite_1900','idem_1900', 'multi_1900','en_sometimes_euph_1900','en_sometimes_euph_no_boundary_1900']


seed = 10

from sklearn.metrics import f1_score, precision_score, recall_score, accuracy_score

table = {'train_data': [], 'category': [], 'average_f1': [], 'average_prec': [], 'average_rec': []}

aa = pd.read_csv('en_euph.csv')

for i in datas:
    for j in aa['CATEGORY'].dropna().unique():
        #print(j)
        total_f1 = 0
        total_prec = 0
        total_rec = 0
        for k in range(0,seed):
            a = pd.read_csv('tables_bert/'+str(k)+'_'+i+'__noboundary_table.csv')
            a = pd.merge(a, aa, how = 'left', left_on = 'id', right_on = 'ID')
            a = a[a['CATEGORY']==j]
            #print(a.head())
            total_f1 += f1_score(a['label'],a['predicted'],average='macro')
            total_prec += precision_score(a['label'],a['predicted'],average='macro')
            total_rec += recall_score(a['label'],a['predicted'],average='macro')
        table['train_data'].append(i)
        table['category'].append(j)
        #print(table['category'])
        table['average_f1'].append(total_f1/seed)
        table['average_prec'].append(total_prec/seed)
        table['average_rec'].append(total_rec/seed)
#print(table)
```

```python
table = pd.DataFrame(table)
table.to_csv('euph_noboundary_1900_bert_category_table.csv')
```
